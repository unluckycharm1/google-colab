```python
df["pass"] = (df["G3"] >= 10).astype(int)
```

```python
print(df["pass"].value_counts())
```

```python
features = [
    "age",
    "studytime",
    "failures",
    "G1",
    "G2"
]
x = df[features]
y = df["pass"]
```

```python
x_train, x_test, y_train, y_test = train_test_split(
    x,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
    )
```

```python
model = DecisionTreeClassifier(
    max_depth=3,
    random_state=42
)
model.fit(x_train, y_train)
```

```python
y_pred = model.predict(x_test)
print(y_pred)
```

```python
accuracy = accuracy_score(y_test, y_pred)
print("Accuracy: ", accuracy)
```

```python
cm = confusion_matrix(y_test, y_pred)

disp = ConfusionMatrixDisplay(
    confusion_matrix=cm,
    display_labels=model.classes_)
disp.plot()
plt.title("Student pass/Fail Confusion Matrix")
plt.show()
```

```python
results = x_test.copy()
results["actual"] = y_test
results["predicted"] = y_pred
results.head(10)
```

```python
results["Risk"] = results["predicted"].map({
    0: "At Risk",
    1: "likely to pass"
})

results.head(10)
```

```python
results["recommendation"] = results["predicted"].map({
    0:"Prioritize for academic intervention",
    1:"continue regular monitoring"
})
results.head(10)
```

```python
probabilities = model.predict_proba(x_test)
results["Probability_of_Failing"] = probabilities[:,0]

results.sort_values(
    "Probability_of_Failing",
    ascending=False
).head(20)
```
